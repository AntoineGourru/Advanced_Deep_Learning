# Exercice 2 — Classification de séries temporelles avec LSTM (7 points)

## Objectif

Vous devez classifier des séries temporelles artificielles en **2 classes**.

Contrairement à une simple classification de fréquence locale, le label dépend ici d’une **relation longue distance** entre un motif placé au début de la séquence et un motif placé vers la fin de la séquence.

Chaque série a la forme **$(60, 1)$**.

## Construction du label

Deux motifs courts sont insérés :
- un motif précoce vers le début de la séquence ;
- un motif tardif vers la fin de la séquence.

La classe dépend de leur relation :
- **classe 0** : les deux motifs ont le même signe ;
- **classe 1** : les deux motifs ont des signes opposés.

Ce choix rend le problème plus adapté à un modèle récurrent, car la décision dépend d’une information située à deux moments éloignés.

## Architecture attendue

Vous devez construire un modèle basé sur une couche LSTM :
- entrée de dimension 1 à chaque pas de temps ;
- représentation interne de dimension 32 ;
- sortie finale de dimension 2.

La façon de transformer la sortie séquentielle du LSTM en une représentation finale de dimension 32 fait partie de l’exercice.

```python
import math
from torch.utils.data import TensorDataset, DataLoader

def generer_serie_longue_dependance(classe, longueur=60):
    serie = np.random.normal(loc=0.0, scale=0.08, size=longueur).astype(np.float32)

    motif = np.array([0.2, 0.7, 1.0, 0.7, 0.2], dtype=np.float32)
    signe_debut = np.random.choice([-1.0, 1.0])

    if classe == 0:
        signe_fin = signe_debut
    else:
        signe_fin = -signe_debut

    # Motif précoce et motif tardif
    serie[5:10] += signe_debut * motif
    serie[45:50] += signe_fin * motif

    # Petit fond sinusoïdal pour éviter une tâche trop triviale
    t = np.linspace(0, 1, longueur)
    serie += 0.05 * np.sin(2 * math.pi * 3 * t).astype(np.float32)

    return serie.astype(np.float32)


def creer_dataset_longue_dependance(n_train=1600, n_test=400, longueur=60):
    X_train, y_train = [], []
    X_test, y_test = [], []

    for _ in range(n_train):
        classe = np.random.randint(0, 2)
        X_train.append(generer_serie_longue_dependance(classe, longueur))
        y_train.append(classe)

    for _ in range(n_test):
        classe = np.random.randint(0, 2)
        X_test.append(generer_serie_longue_dependance(classe, longueur))
        y_test.append(classe)

    X_train = np.array(X_train, dtype=np.float32)[:, :, None]
    X_test = np.array(X_test, dtype=np.float32)[:, :, None]
    y_train = np.array(y_train, dtype=np.int64)
    y_test = np.array(y_test, dtype=np.int64)

    return X_train, y_train, X_test, y_test


X_train_seq, y_train_seq, X_test_seq, y_test_seq = creer_dataset_longue_dependance()

train_dataset_seq = TensorDataset(
    torch.tensor(X_train_seq),
    torch.tensor(y_train_seq)
)

test_dataset_seq = TensorDataset(
    torch.tensor(X_test_seq),
    torch.tensor(y_test_seq)
)

train_loader_seq = DataLoader(train_dataset_seq, batch_size=64, shuffle=True)
test_loader_seq = DataLoader(test_dataset_seq, batch_size=64, shuffle=False)

print("Forme de X_train_seq :", X_train_seq.shape)
print("Forme de y_train_seq :", y_train_seq.shape)
```
