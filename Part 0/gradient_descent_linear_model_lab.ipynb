## 1. Create a very simple dataset

We use a small dataset that approximately follows a straight line.


```python
import numpy as np
import matplotlib.pyplot as plt

# Simple training data
x = np.array([0., 1., 2., 3., 4.])
y = np.array([1.1, 2.9, 5.2, 6.8, 9.1])

plt.scatter(x, y)
plt.xlabel("x")
plt.ylabel("y")
plt.title("Training data")
plt.show()
```

## 2. Define the model

Our model is

$$\hat{y} = wx + b,$$

where $w$ is the slope and $b$ is the intercept. We start with $w=0$ and $b=0$.


```python
w = 0.0
b = 0.0

def predict(x, w, b):
    return w * x + b

y_pred = predict(x, w, b)
print("Initial predictions:", y_pred)
```

## 3. Define the loss

We use mean squared error:

$$L = \frac{1}{N} \sum_{i=1}^{N}(\hat{y}_i-y_i)^2.$$

A smaller loss means that the predictions are closer to the observed values.


```python
def mse(y, y_pred):
    return np.mean((y_pred - y) ** 2)

initial_loss = mse(y, y_pred)
print("Initial loss:", initial_loss)
```

## 4. Compute the gradients

For MSE and the model $\hat{y}=wx+b$, the gradients are

$$\frac{\partial L}{\partial w} = \frac{2}{N}\sum_i (\hat{y}_i-y_i)x_i,$$

$$\frac{\partial L}{\partial b} = \frac{2}{N}\sum_i (\hat{y}_i-y_i).$$

Gradient descent updates the parameters in the opposite direction of the gradient:

$$w \leftarrow w - \eta \frac{\partial L}{\partial w}, \qquad b \leftarrow b - \eta \frac{\partial L}{\partial b},$$

where $\eta$ is the learning rate.


```python
def gradients(x, y, w, b):
    y_pred = predict(x, w, b)
    error = y_pred - y

    dw = 2 * np.mean(error * x)
    db = 2 * np.mean(error)
    return dw, db

dw, db = gradients(x, y, w, b)
print("dw =", dw)
print("db =", db)
```

## 5. Perform one gradient descent step

Run the next cell several times if you want to see how a single update changes the parameters and loss.


```python
learning_rate = 0.05

dw, db = gradients(x, y, w, b)
w = w - learning_rate * dw
b = b - learning_rate * db

y_pred = predict(x, w, b)

print(f"w = {w:.4f}")
print(f"b = {b:.4f}")
print(f"loss = {mse(y, y_pred):.4f}")
```

## 6. Train the model

Now repeat the update for many iterations. We store the loss so that we can inspect the optimization process.


```python
# Reset the parameters
w = 0.0
b = 0.0

learning_rate = 0.05
num_steps = 100
loss_history = []

for step in range(num_steps):
    # Forward pass
    y_pred = predict(x, w, b)
    loss = mse(y, y_pred)
    loss_history.append(loss)

    # Compute gradients
    dw, db = gradients(x, y, w, b)

    # Gradient descent update
    w = w - learning_rate * dw
    b = b - learning_rate * db

    if step % 10 == 0:
        print(f"step {step:3d} | loss = {loss:.4f} | w = {w:.4f} | b = {b:.4f}")

print("\nLearned parameters")
print(f"w = {w:.4f}")
print(f"b = {b:.4f}")
```

## 7. Plot the training loss

The loss should generally decrease as gradient descent improves the parameters.


```python
plt.plot(loss_history)
plt.xlabel("Gradient descent step")
plt.ylabel("MSE loss")
plt.title("Training loss")
plt.show()
```

## 8. Plot the learned model

Compare the learned line with the training observations.


```python
x_line = np.linspace(x.min(), x.max(), 100)
y_line = predict(x_line, w, b)

plt.scatter(x, y, label="data")
plt.plot(x_line, y_line, label="learned model")
plt.xlabel("x")
plt.ylabel("y")
plt.title("Linear model learned with gradient descent")
plt.legend()
plt.show()
```
